Frontend: NiceGUI.
Backend (lógica): Python.
ORM: SQLAlchemy.
Base de datos: SQLite (con posibilidad de migrar a PostgreSQL si el sistema crece).
Reportes: Pandas + OpenPyXL para Excel, y una librería como ReportLab o WeasyPrint para PDF.

Para un proyecto personal, una pila bastante potente sería:

NiceGUI
SQLAlchemy
Alembic
Pydantic
Jinja2
WeasyPrint
Pandas
APScheduler
SQLite → PostgreSQL
Docker

Con eso puedes hacer aplicaciones administrativas completas y, al mismo tiempo, aprender conceptos que luego se transfieren casi directamente a Django.
